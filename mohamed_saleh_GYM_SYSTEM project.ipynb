```python
from datetime import date, timedelta


class GYM_System:

    def __init__(self):
        self.name = input("please enter your name: ")
        self.gender = input("please enter your gender: ")
        self.age = int(input("please enter your age: "))
        self.ID = input("please enter your ID: ")
        self.__locker_password = input("please enter your locker password: ")

    def member_info(self):
        print("\nMember Information")
        print("------------------")
        print("Name:", self.name)
        print("Gender:", self.gender)
        print("Age:", self.age)
        print("ID:", self.ID)

    def subscription(self):
        print("\n1. 1 Month")
        print("2. 3 Months")
        print("3. 6 Months")
        print("4. 1 Year")

        choice = input("Choose subscription: ")

        if choice == "1":
            days = 30
        elif choice == "2":
            days = 90
        elif choice == "3":
            days = 180
        elif choice == "4":
            days = 365
        else:
            print("Invalid choice")
            return

        self.start_date = date.today()
        self.end_date = self.start_date + timedelta(days=days)

        print("\nSubscription Information")
        print("Member:", self.name)
        print("Start Date:", self.start_date)
        print("End Date:", self.end_date)

    def renew_subscription(self):

        print("\n1. 1 Month")
        print("2. 3 Months")
        print("3. 6 Months")
        print("4. 1 Year")

        choice = input("Choose renewal: ")

        if choice == "1":
            days = 30
        elif choice == "2":
            days = 90
        elif choice == "3":
            days = 180
        elif choice == "4":
            days = 365
        else:
            print("Invalid choice")
            return

        self.end_date = self.end_date + timedelta(days=days)

        print("Subscription renewed successfully")
        print("New End Date:", self.end_date)

    def gym_coach_info(self):
        print("\nTrainer Information")
        print("--------------------")
        print("Name: Cap Salah")
        print("Age: 37")
        print("Phone: 01012345678")
        print("Specialization: Body Building")
        print("Experience: 20 years")

    def workout_plan(self):
        print("\nWorkout Plan")
        print("------------")
        print("Saturday: Chest + Triceps")
        print("Sunday: Back + Biceps")
        print("Monday: Rest")
        print("Tuesday: Shoulders")
        print("Wednesday: Legs")
        print("Thursday: Cardio")
        print("Friday: Rest")

    def check_subscription(self):
        today = date.today()

        if today <= self.end_date:
            remaining = (self.end_date - today).days
            print("\nSubscription is active")
            print("Remaining days:", remaining)
            print("End Date:", self.end_date)
        else:
            print("\nSubscription expired")

    def change_locker_password(self):
        old_pass = input("Please enter your old password: ")

        if old_pass == self.__locker_password:
            new_pass = input("Please enter your new password: ")
            self.__locker_password = new_pass
            print("Password changed successfully")
        else:
            print("Wrong password")
```

```python
gym = GYM_System()

gym.member_info()

gym.subscription()

gym.check_subscription()

gym.renew_subscription()

gym.check_subscription()

gym.gym_coach_info()

gym.workout_plan()

gym.change_locker_password()
```
